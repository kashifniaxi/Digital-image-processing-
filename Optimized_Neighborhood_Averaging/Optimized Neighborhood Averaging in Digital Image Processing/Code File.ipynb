```python
import numpy as np

# Create an 11x11 image with sequential pixel values
image = np.arange(1, 122).reshape(11, 11)
print("11x11 Image:\n", image)
```

```python
def traditional_averaging(image, mask_size):
    import numpy as np

    pad_size = mask_size // 2
    padded_image = np.pad(image, ((0, 0), (pad_size, pad_size)), mode='constant', constant_values=0)
    center_row = padded_image[5, :]  # Center row (zero-padded)

    output_row = []
    operation_count = {'additions': 0, 'multiplications': 0, 'divisions': 0}

    for i in range(len(image[0])):
        # Extract the window
        window = center_row[i:i + mask_size]
        # Perform averaging
        sum_pixels = sum(window)
        operation_count['additions'] += mask_size - 1
        average = sum_pixels / (mask_size ** 2)
        operation_count['divisions'] += 1
        output_row.append(average)

    return np.array(output_row), operation_count

image = np.arange(1, 122).reshape(11, 11)
output, operations = traditional_averaging(image, mask_size=5)
print("Output Row (Traditional):", output)
print("Operations (Traditional):", operations)
```

```python
def optimized_averaging(image, mask_size):
    import numpy as np

    pad_size = mask_size // 2
    padded_image = np.pad(image, ((0, 0), (pad_size, pad_size)), mode='constant', constant_values=0)
    center_row = padded_image[5, :]  # Center row (zero-padded)

    output_row = []
    operation_count = {'additions': 0, 'subtractions': 0, 'divisions': 0}

    # Calculate the first window
    window = center_row[:mask_size]
    sum_pixels = sum(window)
    operation_count['additions'] += mask_size - 1
    output_row.append(sum_pixels / (mask_size ** 2))
    operation_count['divisions'] += 1

    # Slide the window across
    for i in range(1, len(image[0])):
        sum_pixels += center_row[i + mask_size - 1] - center_row[i - 1]
        operation_count['additions'] += 1
        operation_count['subtractions'] += 1
        average = sum_pixels / (mask_size ** 2)
        operation_count['divisions'] += 1
        output_row.append(average)

    return np.array(output_row), operation_count

output_optimized, operations_optimized = optimized_averaging(image, mask_size=5)
print("Output Row (Optimized):", output_optimized)
print("Operations (Optimized):", operations_optimized)
```
