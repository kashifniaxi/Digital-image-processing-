```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
 
```

```python
def plot_bit_images(encrypted_image,decrypted_image,i):
    plt.figure(figsize=(12, 6))

    # Displaying the encrypted image
    plt.subplot(1, 2, 1)
    plt.title(f'Encrypted Image with i={i}')
    plt.imshow(encrypted_image, cmap='gray')
    plt.axis('off')

    # Displaying the decrypted image
    plt.subplot(1, 2, 2)
    plt.title(f'Decrypted Image with i={i}')
    plt.imshow(decrypted_image, cmap='gray')
    plt.axis('off')

    plt.show()
```

```python
def load_image(url):
    image = cv2.imread(url,cv2.IMREAD_GRAYSCALE)
    return image
def display_image(image):
    plt.imshow(image,cmap='gray')
    plt.axis('off')  # Turn off axis numbers and ticks
    plt.show()
cover= load_image(r'/home/umerfarooq/DIP/DIP/horse.jpg')
msg= load_image(r'/home/umerfarooq/DIP/DIP/sea.jpg')
display_image(cover)
display_image(msg)
```

```python
def eightBitPlane(image):
    plt.figure(figsize=(12, 8))
    
    for i, bit in enumerate([128, 64, 32, 16, 8, 4, 2, 1]):
        # Isolate the bit plane
        bit_plane = (image & bit) >> (7 - i)  # Shift to the right to align it
        plt.subplot(2, 4, i + 1)  # Create a grid for the images
        plt.imshow(bit_plane, cmap='gray')
        plt.title(f' {i+1} Bit Plane')
        plt.axis('off')

    plt.tight_layout()
    plt.show()
eightBitPlane(cover)
```

```python
def stegnography(coverimage,msgimage,bit):
    coverimage = np.array(coverimage,np.uint8)
    msgimage = np.array(msgimage,np.uint8)
    coverimage = coverimage>>bit
    coverimage <<=bit
    msgimage>>=8-bit
    encryptedimage = coverimage | msgimage
    # display_image(encryptedimage)
    return encryptedimage

def decryptmsg(image,bit):
    decryptedimage = (image & ((1 << bit) - 1)) << (8 - bit)
    # display_image(decryptedimage)
    return decryptedimage
bits = 1
encryptedimage = stegnography(cover,msg,bits)
decryptedimage = decryptmsg(encryptedimage,bits)
plot_bit_images(encryptedimage,decryptedimage,bits)
```

```python
for i in range(2,8):
    encryptedimage = stegnography(cover,msg,i)
    decryptedimage = decryptmsg(encryptedimage,i)
    plt.figure(figsize=(12, 6))

    # Displaying the encrypted image
    plt.subplot(1, 2, 1)
    plt.title(f'Encrypted Image with i={i}')
    plt.imshow(encryptedimage, cmap='gray')
    plt.axis('off')

    # Displaying the decrypted image
    plt.subplot(1, 2, 2)
    plt.title(f'Decrypted Image with i={i}')
    plt.imshow(decryptedimage, cmap='gray')
    plt.axis('off')

    plt.show()
```
